```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import expit  # sigmoid for smooth saturation
```

# Concept Image - Figure 1

```python

# f1(x1 | x3)
def f11(x1):
    return 1.2 * x1**3 + 0.1 * x1  # steeper growth early

def f12(x1):
    return 0.9 * np.tanh(2 * x1)  # saturates later

def f1(x1, x3):
    y = np.zeros_like(x1)

    cond = x3 > 0
    y[cond] = f11(x1[cond])
    y[~cond] = f11(x1[~cond])
    return y

def f21(x2):
    return 0.4 * np.sin(np.pi * x2 / 2 - 0.2) - 0.4  # same shape, lower offset

def f22(x2):
    return 0.6 * np.sin(np.pi * x2 / 2 - 0.1) - 0.2  # same shape, lower offset

def f23(x2):
    return 0.8 * np.sin(np.pi * x2 / 2)  # natural bell-like response

def f2(x2, x1):
    y = np.zeros_like(x2)

    cond1 = x1 < - 0.4
    cond2 = np.logical_and(x1 >= -0.4, x1 < 0.4 )
    cond3 = x1 >= 0.4
    y[cond1] = f21(x2[cond1])
    y[cond2] = f22(x2[cond2])
    y[cond3] = f23(x2[cond3])
    return y


def f3 (x3):
    return  1.5 * (expit(2 * x3) - 0.5)  # maps x3 in [-1,1] to roughly [-0.75, 0.75]

x = np.linspace(-1, 1, 200)
folder = "test_synth_user_study/"
import os 
os.makedirs(folder, exist_ok=True)
import matplotlib.ticker as ticker
fontsize=17
# --- Figure 1: f1(x1 | x3) ---
plt.figure()
plt.plot(x, f11(x), label=r"$f(x_1 \mid x_3 > 0)$")
plt.plot(x, f12(x), label=r"$f(x_1 \mid x_3 \leq 0)$")

plt.axvline(x=-0.4, color='gray', linestyle=':')
plt.axvline(x=0.4, color='gray', linestyle=':')
plt.text(-0.4, -0.9, r'$x_2\ (\uparrow):[0,0.37]$', ha='center', va='top', fontsize=fontsize)
plt.text(0.4, -0.9, r'$x_2\ (\uparrow):[0,0.37]$', ha='center', va='top', fontsize=fontsize)

plt.xticks([])
plt.yticks([])
# plt.ylim(-1.5, 1.5)
plt.legend(fontsize=fontsize,)
plt.tight_layout()
ax = plt.gca()
for spine in ax.spines.values():
    spine.set_visible(False)

plt.show()

# --- Figure 2: f2(x2 | x1) ---
plt.figure()
plt.plot(x, f21(x), label=r"$f(x_2 \mid x_1 < -0.4)$")
plt.plot(x, f22(x), label=r"$f(x_2 \mid x_1 \in [-0.4, 0.4])$")
plt.plot(x, f23(x), label=r"$f(x_2 \mid x_1 \geq 0.4)$")

plt.xticks([])
plt.yticks([])
plt.legend(fontsize=fontsize,)
plt.tight_layout()
ax = plt.gca()
for spine in ax.spines.values():
    spine.set_visible(False)

ax = plt.gca()


plt.show()

# --- Figure 3: f3(x3) ---
plt.figure()
plt.plot(x, f3(x), label=r"$f_d(x_d)$")
plt.axvline(x=0, color='gray', linestyle=':')
plt.text(0, -0.4, r'$x_1\ (\updownarrow)[-0.42,0,42]$', ha='center', va='top', fontsize=fontsize)

plt.xticks([])
plt.yticks([])
# plt.ylim(-1.5, 1.5)
plt.legend(fontsize=fontsize,)
plt.tight_layout()
ax = plt.gca()
for spine in ax.spines.values():
    spine.set_visible(False)

plt.show()
```

The regions are defined conditioning on the interacting features:
* Τhe effect of x1 conditions on x3 (Cx3 )
* he effect of x2 conditions on x1 (Cx1 )
* xd does not interact with any other feature and thus has a single plot

# Figure 2: CALM plot for x1

```python
fontsize = 15
# --- Figure 1: f1(x1 | x3) ---
plt.figure()
plt.plot(x, f11(x), label=r"$f(x_1 \mid x_3 > 0)$")
plt.plot(x, f12(x), label=r"$f(x_1 \mid x_3 \leq 0)$")

plt.axvline(x=-0.4, color='gray', linestyle=':')
plt.axvline(x=0.4, color='gray', linestyle=':')
plt.text(-0.4, -0.9, r'$x_2\ (\uparrow):[0,0.37]$', ha='center', va='top', fontsize=fontsize)
plt.text(0.4, -0.9, r'$x_2\ (\uparrow):[0,0.37]$', ha='center', va='top', fontsize=fontsize)

plt.xlabel(r"$x_1$", fontsize=fontsize)
plt.ylabel(r"$y$", fontsize=fontsize)
plt.xticks([-1, -.5, 0, 0.5, 1])
plt.yticks([-1.5, -.75, 0, .75, 1.5])
plt.legend(fontsize=fontsize)
plt.tight_layout()

ax = plt.gca()
ax.tick_params(axis='both', labelsize=fontsize-1)

plt.show()
```

Each curve gives the contribution of $ x_1 $ to $ y $ in a specific region.
* The blue curve when $ x_3 > 0 $ 
* The orange curve when $ x_3 \leq 0 $

For example, at $ x_1 = -0.5 $, the contribution is approximately $-0.2$ (blue) or $-0.75$ (orange), depending on $ x_3 $.
The plots also illustrate how altering $ x_1 $ to $x_1 \rightarrow x_1 + \delta$ impacts the prediction.
Vertical dotted lines mark points of a hidden discontinuity which is due to $x_1$ participating as an interaction term for feature $x_2$. As shown in previous image, the effect of $ x_2 $ is conditioned by $x_1 \leq - 0.4 $, $-0.4 \leq x_1 \leq 0.4 $ and $x_1 > 0.4$, therefore in this figurer we observe vertical lines in $x_1 \pm 0.4$.
If a change in $x_1$ does not cross a vertical line, the change in the output $ (\Delta y) $ equals the curve difference $ (\Delta f_i )$.
Crossing a line signifies a hidden jump, in the range $[\alpha, \beta]$, so 
$ \Delta f_i + a \leq \Delta y \leq \Delta f_i + \beta$.

Arrows provide a fast understanding of the jump:
* $\uparrow$ means $\Delta y > \Delta f_i$,
* $\downarrow$ means $\Delta y < \Delta f_i$,
* $\updownarrow$ means it depends.
